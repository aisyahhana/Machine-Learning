```python
# Import library dasar
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D # untuk 3D plot
# Agar plot tampil rapi
plt.rcParams['figure.figsize' ] = (8, 6)
```

```python
df = pd.read_csv('https://drive.google.com/uc?export=download&id=1DTmhHBxhFdqpAPYQQa4KdIVrgS6XCOAA')
```

```python
df.info()
```

```python
df = df.drop(['id', 'Unnamed: 32'], axis=1)
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
X = df.drop('diagnosis', axis=1)
y = df['diagnosis']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

```python
explained_variance = pca.explained_variance_ratio_

plt.figure()
plt.plot(np.cumsum(explained_variance))
plt.xlabel('Jumlah Komponen')
plt.ylabel('Variansi Kumulatif')
plt.title('Explained Variance oleh PCA')
plt.show()
```

```python
pca = PCA(n_components=2)
X_pca_2 = pca.fit_transform(X_scaled)
```

```python
plt.figure()
plt.scatter(X_pca_2[:, 0], X_pca_2[:, 1], c=y)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Visualisasi PCA Dataset Medis')
plt.show()
```
